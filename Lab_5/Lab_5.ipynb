### Task#1: Convolve an image with a 7x7 box filter, and output the original and smoothed image side by side

```python
import cv2
import matplotlib.pyplot as plt
from skimage import data

# Use skimage's coffee image.
original = data.coffee()

# 7x7 box filter.
smoothed = cv2.blur(original, (7, 7))

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(original)
plt.title('Original Coffee Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(smoothed)
plt.title('7x7 Box Filter Output')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Task#2: Using an image of your choice, apply a 5x5 Gaussian filter to the image, then a 21x21 Gaussian filter, and show the three images side by side.

```python
import cv2
import matplotlib.pyplot as plt
from skimage import data

# Use skimage's kidney image.
image = data.camera()

# Gaussian smoothing with two different kernel sizes.
gaussian_5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21 = cv2.GaussianBlur(image, (21, 21), 0)

plt.figure(figsize=(15, 5))
images = [image, gaussian_5, gaussian_21]
titles = ['Original Kidney Image', '5x5 Gaussian Filter', '21x21 Gaussian Filter']

for i, (img, title) in enumerate(zip(images, titles), start=1):
    plt.subplot(1, 3, i)
    if img.ndim == 2:
        plt.imshow(img, cmap='gray')
    else:
        plt.imshow(img)
    plt.title(title)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

### Task#3: Using an image of your choice, apply a 3x3 Laplacian filter to sharpen the image. Follow the following steps:
1.	Read the image
2.	Convert image to float32
3.	Use the cv2.Laplacian() function
4.	Sharpen the image using: np.clip(image_float - laplacian, 0, 1)
5.	Plot the original image, the resulting image after applying the Laplacian filter, and the sharpened image


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

# 1. Read the image.
image = data.horse().astype(np.uint8) * 255

# 2. Convert image to float32.
image_float = image.astype(np.float32) / 255.0

# 3. Use the cv2.Laplacian() function.
laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

# 4. Sharpen the image.
sharpened = np.clip(image_float - laplacian, 0, 1)

# 5. Plot the original image, Laplacian result, and sharpened image.
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image_float, cmap='gray')
plt.title('Original Horse Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Filter Output')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sharpened, cmap='gray')
plt.title('Sharpened Image')
plt.axis('off')

plt.tight_layout()
plt.show()
```
